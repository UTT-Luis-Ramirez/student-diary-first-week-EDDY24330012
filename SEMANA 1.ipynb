#ESTADISTICA DIFERENCIAL
DIFERENCIAL: RECOLECCION Y ORGANIZACION DE DATOS, CALCULO DE MEDIDAS
DESCRIPTIVAS, PRESENTACION DE DATOS, IDENTIFICACION DE PATRONES Y RELACIONES
TENDENCIA CENTRAL: MEDIA, MEDIANA, MODA, MEDIA GEOMETRICA

DISPERSIÓN: GRADO DE VARIABILIDAD O DISTRIBUCIÓN DE LOS DATOS EN UN CONJUNTO. ES DECIR, MIDE CÓMO SE DISPERSAN LOS DATOS ALREDEDOR DE UNA MEDIDA DE TENDENCIA CENTRAL COMO LA MEDIA.

TENDENCIA CENTRAL:SE REFIERE A VALORES QUE REPRESENTAN UN CONJUNTO DE DATOS MEDIANTE UN ÚNICO VALOR CENTRAL. ESTAS MEDIDAS NOS AYUDAN A ENTENDER DÓNDE SE ENCUENTRAN LOS DATOS EN PROMEDIO O MÁS FRECUENTEMENTE. LAS TRES MEDIDAS MÁS COMUNES DE TENDENCIA CENTRAL SON MEDIA MEDIANA Y MODA

TAREA 2
Control de calidad del software:

Ejemplo práctico:
Durante las pruebas de un software de facturación, se registraron 25 defectos distribuidos en tres módulos:
Módulo de Cálculos: 15 defectos.
Módulo de Reportes: 7 defectos.
Módulo de Exportación: 3 defectos.
Conclusión:* Se priorizó la corrección del módulo de cálculos, lo que redujo defectos en un 60%

DE LOS 3 MÓDULOS, EL MÓDULO DE CÁLCULOS REPRESENTA MÁS DE LA MITAD DE LOS MÓDULOS, Y ESTO HACE QUE LOS DEFECTOS BAJEN UN 60%, TENIENDO QUE PRIORIZAR MÁS A ESTE PRIMER MÓDULO POR REPRESENTAR MÁS DEFECTOS QUE LOS DEMÁS,ESE 60% BAJARÍA TAMBIÉN A LOS DEMÁS, YA QUE TAL VEZ LOS DEMÁS MÓDULOS ESTÉN TENIENDO DEFECTOS A CAUSA DE ESTE PRIMERO…TIENE MUCHO SENTIDO, ESE 60 DARÍA AL POSIBLE 100, YA QUE ESTE ES EL ÚNICO QUE FALLA MUCHO
